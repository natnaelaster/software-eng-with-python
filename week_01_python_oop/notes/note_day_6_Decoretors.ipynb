<b> Concept 1: functions are objects (the prerequisite for decorators)</b>

<h6>
The analogy. A function is like a recipe card. You can read it, copy it, hand it to someone, or put it in a drawer next to other cards. 

Calling the function is cooking the recipe. Most beginners only ever cook. Today you start passing the card around.

This is a foundational concept. You've used functions. Today you learn to treat them like data — pass them, store them, return them.

<b> Part 1: The Core Idea</b>

A function in Python is an object. You can:

- Assign it to a variable

- Store it in a list or dict

- Pass it as an argument to another function

- Return it from a function

This is what makes decorators possible. Without it, none of the later concepts work.

<strong> Part 1: A function is a value </strong>

`def shout(text):` creates a function object and binds the name `shout` to it.

The name `shout` **is** the object. Adding parentheses (`shout()`) **calls** it. Without parentheses, you're holding the object itself.

That's the same as `x = 5` — `x` is a name bound to an int. `shout` is a name bound to a function. Both are values. You can reassign them, pass them, store them.

Three things you can do once functions are objects:

1. **Assign** to another name: `yell = shout`
2. **Store** in a data structure: `actions = {"loud": shout}`
3. **Pass** into another function: `apply_twice(shout, "hey")`

Each one is demonstrated below.

```python
def shout(text):
    return text.upper() + "!"

# 1. Assign to another name (no parentheses = no call)
yell = shout
print(yell("hello"))            # HELLO!

# 2. Store in a data structure
actions = {"loud": shout, "plain": str}
print(actions["loud"]("hi"))    # HI!

# 3. Pass INTO another function
def apply_twice(func, value):
    return func(func(value))

print(apply_twice(shout, "hey"))   # HEY!!
```

<h6>

<strong> Line-by-line </strong>

`yell = shout` — no parentheses on either side. `yell` is now a second name for the **same** function object. `yell is shout` is `True`.

`actions = {"loud": shout, "plain": str}` — functions stored as dict values. `actions["loud"]` returns the function; adding `("hi")` calls it.

`apply_twice(shout, "hey")` — `func` is bound to the `shout` function object. Inside, `func("hey")` → `"HEY!"`, then `func("HEY!")` → `"HEY!!"`.

Notice: `shout` was written **without** parentheses when passed to `apply_twice`. If you write `shout()`, Python calls it immediately and passes the **result** — not the function.

<strong> What about the expected output </strong>

The notes said `HEY!!!` (three exclamation marks). That was a typo in the original concept text — the actual output is `HEY!!` (two). This is why you **always run the code** and trust the actual output over the notes.

<strong> Part 2: Closures — a function that remembers </strong>

A function defined **inside** another function can use the outer function's variables, even after the outer function has finished.

This is called a **closure**: the inner function "closes over" the outer variables, capturing them for later use.

Why this works: when you call `make_multiplier(2)`, the local `factor` would normally be destroyed when the function returns. But because `multiply` still references it, 

Python keeps the variable alive — attached to the returned function.

Each call to `make_multiplier` creates a **new, independent** closure. `double` remembers `factor = 2`. `triple` remembers `factor = 3`. They do not share state.

</h6>

```python
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))   # 20
print(triple(10))   # 30
```

<h6>

<strong> Line-by-line </strong>

`def make_multiplier(factor):` — outer function, takes `factor`.

`def multiply(x):` — inner function, defined inside `make_multiplier`.

`return x * factor` — reads `factor` from the enclosing scope. This is what makes it a closure.

`return multiply` — returns the inner **function object** (no parentheses → no call). If you wrote `return multiply()`, Python would call it and try to return the result — which would fail because `multiply` needs an argument.

`double = make_multiplier(2)` — runs the outer function once. `factor = 2`. Returns the `multiply` function with `factor` captured. Now `double(10)` reads `factor = 2` → `20`.

`triple = make_multiplier(3)` — a **separate** call. Fresh `factor = 3`. Returns a **different** `multiply` function. `triple(10)` reads `factor = 3` → `30`.

<strong> **Concept 2: decorators — wrapping a function without touching it**</strong>

<h6>
<strong> The analogy </strong>

A security checkpoint at an airport doesn't change you or where you're going. It wraps your trip with extra steps before and after. A decorator does the same to a function: same function, same result, extra behavior around it.
</h6>

<h6>
A decorator is a function that <b>takes a function</b> and <b>returns a function</b>. That's the whole definition.

`timed(func)` accepts a function, defines a new inner function `wrapper` that does extra work before and after calling `func`, and returns `wrapper`. The original `func` is untouched. The new function behaves the same but with extra behavior wrapped around it.

The reassignment `slow_sum = timed(slow_sum)` is the entire mechanism. You're not modifying `slow_sum`. You're replacing the name `slow_sum` with a new function object (the wrapper) that internally calls the old `slow_sum`.
</h6>

```python
import time

def timed(func):
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"{func.__name__} took {elapsed:.4f}s")
        return result
    return wrapper

def slow_sum(n):
    return sum(range(n))

slow_sum = timed(slow_sum)
slow_sum(10_000_000)
```

<h6>
<strong> Line-by-line </strong>

`def timed(func):` — takes a function as its only argument. `func` is bound to the function object passed in.

`def wrapper(*args, **kwargs):` — defines an inner function. `*args` collects positional arguments into a tuple. `**kwargs` collects keyword arguments into a dict. Together they mean <b>`"accept any arguments the original function might take."`</b>

`start = time.perf_counter()` — records the time before calling `func`.

`result = func(*args, **kwargs)` — calls the original function. `*args` unpacks the tuple back into positional arguments. `**kwargs` unpacks the dict back into keyword arguments. This is how the wrapper forwards any signature unchanged.

`elapsed = time.perf_counter() - start` — measures the duration.

`print(...)` — extra behavior after the call.

`return result` — passes the original function's return value through. <b>If you forget this line, the decorated function silently returns `None`.</b> This is the most common decorator bug.

`return wrapper` — returns the new function object. No parentheses — we're not calling it, we're handing it back.

`slow_sum = timed(slow_sum)` — rebinds the name `slow_sum` to the wrapper. The original function is still alive (captured by `wrapper`), but no longer accessible by name.
</h6>

<h6>
<b> The `@` syntax </b>

`@` is shorthand for the reassignment. Nothing more.
</h6>

```python
@timed
def slow_sum(n):
    return sum(range(n))

"""is exactly equivalent to:"""

def slow_sum(n):
    return sum(range(n))

slow_sum = timed(slow_sum)
```

<h6>
The `@` line is read as "apply the decorator `timed` to the function defined immediately below, and rebind the name to the result."

<b>Order matters:</b> the decorator function is <b>called</b> at the moment of definition. `timed(slow_sum)` runs as soon as Python sees the `@timed` block. The wrapper is created. The name `slow_sum` is rebound. Then, later, when you call `slow_sum(...)`, you're calling the wrapper.
</h6>

<h6>
<b> `@wraps` — preserving function identity </b>

Without `@wraps`:
</h6>

```python
def timed(func):
    def wrapper(*args, **kwargs):
        ...
    return wrapper

@timed
def slow_sum(n):
    return sum(range(n))

print(slow_sum.__name__)   # "wrapper"  ← wrong
print(slow_sum.__doc__)    # None       ← wrong
```

<h6>
The decorated function's name is `"wrapper"`, not `"slow_sum"`. Its docstring is `None`, not the original. Why? Because `slow_sum` is now bound to the `wrapper` function, and `wrapper`'s name is literally `"wrapper"`.

This breaks:

- Debuggers — stack traces show `wrapper` instead of the real function name.
- Logs — `f"{func.__name__}"` prints the wrong name.
- Tools like LangChain — they read `__name__` and `__doc__` to register tools. Wrong identity, wrong registration.
- Introspection — `help(slow_sum)` shows nothing useful.

`@wraps` fixes this. It copies the original function's `__name__`, `__doc__`, `__module__`, `__qualname__`, and other metadata onto the wrapper.
</h6>

```python
from functools import wraps

def timed(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        ...
    return wrapper
```

<h6>
Now `slow_sum.__name__` is `"slow_sum"`. Debuggers, logs, and tools see the right identity.

<b>Rule:</b> every decorator you write uses `@wraps(func)` on the inner `wrapper`. No exceptions.
</h6>

<h6>
<b> A decorator with arguments — the three-layer pattern </b>

`@retry(3)` is <b>not</b> a decorator. It's a <b>function that returns a decorator</b>. That's why it needs three layers instead of two.
</h6>

```python
def retry(times):                          # layer 1: receives the argument
    def decorator(func):                   # layer 2: receives the function
        @wraps(func)
        def wrapper(*args, **kwargs):      # layer 3: receives the call
            last_error = None
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except ConnectionError as e:     # specific, as learned on Day 5
                    last_error = e
                    print(f"attempt {attempt}/{times} failed: {e}")
            raise last_error               # all attempts failed: re-raise
        return wrapper
    return decorator

@retry(3)                                  # retry(3) runs first, returns decorator,
def call_llm(prompt):                      # decorator then wraps call_llm
    ...
```

<h6>
<b>Layer-by-layer:</b>

`retry(times)` — runs <b>first</b>, when the `@retry(3)` line is evaluated. Returns `decorator`.

`decorator(func)` — runs <b>second</b>, when Python applies the decorator to the function below. Returns `wrapper`.

`wrapper(*args, **kwargs)` — runs <b>each time</b> the decorated function is called. Does the actual work.

Compare to a no-argument decorator (`@timed`), which has <b>two</b> layers: `timed(func)` → `wrapper(*args, **kwargs)`. The extra layer appears only when the decorator itself takes an argument.

<b>Why the retry pattern matters:</b> production code that calls LLM APIs, web services, or databases has to survive transient failures. Network blips, rate limits, timeouts — these happen. A `retry` decorator wraps every API call. This is one of the most important real-world decorator patterns.
</h6>

<strong> **Concept 3: generators — producing values one at a time**</strong>

<h6>
<strong> The analogy </strong>

A bakery can bake 10,000 loaves and stack them in the shop before opening, or it can bake one loaf whenever a customer asks. The first needs enormous shelf space and a long wait before the first sale. The second serves the first customer instantly and uses almost no space. A list is the first bakery. A generator is the second.
</h6>

<h6>
A normal function computes everything, stores the result, and returns once. If it produces 10 million values, all 10 million exist in memory.

A generator function uses <b>yield</b> instead of <b>return</b>. It produces one value, pauses, and resumes when the next value is requested. Between calls, the function is frozen — its local variables stay alive, its position in the loop is remembered.

Calling a generator function does not run the body. It returns a generator object. The body only runs when something asks for the next value, either <b>next()</b> or a <b>for</b> loop.
</h6>



```python
# A regular function: computes everything, returns once
def squares_list(n):
    result = []
    for i in range(n):
        result.append(i * i)
    return result              # all n values exist in memory at once

# A generator function: `yield` instead of `return`
def squares_gen(n):
    for i in range(n):
        yield i * i            # hand back ONE value, then pause here

gen = squares_gen(3)
print(next(gen))   # 0
print(next(gen))   # 1
print(next(gen))   # 4
print(next(gen))   # raises StopIteration
```

<h6>
<strong> Line-by-line </strong>

`def squares_list(n):` — normal function. Builds a full list, returns it once. All n values exist in memory at the moment of return.

`def squares_gen(n):` — generator function. The presence of `yield` anywhere in the body makes it a generator.

`for i in range(n):` — same loop as the list version.

`yield i * i` — the core difference. Instead of appending to a list and continuing, `yield` pauses the function, hands back the value, and waits. The next call to `next()` resumes execution immediately after the `yield` — same `i`, next iteration.

`gen = squares_gen(3)` — calls the function, but does not run the body. Returns a generator object.

`next(gen)` — first call. Body runs from the top: `i = 0`, `yield 0 * 0` → returns `0`. Function pauses.

`next(gen)` — second call. Resumes right after the first `yield`. Loop continues: `i = 1`, `yield 1 * 1` → returns `1`. Pauses again.

`next(gen)` — third call. `i = 2`, `yield 4`. Pauses.

`next(gen)` — fourth call. Loop is done. Function reaches the end of its body. Raises `StopIteration` to signal "no more values."
</h6>

```python
for value in squares_gen(5):
    print(value)
```

<h6>
The `for` loop calls `next()` for you and stops when `StopIteration` is raised. Same behavior as the manual `next()` calls, but cleaner. This is the idiomatic way to consume a generator — you rarely call `next()` directly outside of specific use cases.
</h6>

<h6>
<b> Complexity </b>

The list version is <b>O(n) memory</b> — it holds all n values at once. The generator is <b>O(1) memory</b> — it holds one value at a time.

Time is <b>O(n)</b> for both if you consume everything. The difference is memory. For 50 million rows, the list version can crash your program. The generator version runs in a few megabytes.

<b> The counterintuitive fact </b>

A generator can only be consumed once. After it's exhausted, it stays empty. If you need to iterate twice, either create a new generator or convert it to a list with <b>list(gen)</b>.
</h6>

<h6>
<b> Generator expressions — the one-line version </b>

For simple cases, a generator expression is a compact form. Same parentheses-as-brackets trick you've seen before, but with round parentheses and no list ever built.
</h6>

```python
total = sum(x * x for x in range(10_000_000))  # no list ever exists in memory
```

<h6>
The syntax: <b>(expression for item in iterable)</b>. The parentheses are the giveaway. Compare:

<b>List comprehension:</b> <b>[x * x for x in range(10)]</b> — builds a list, all values in memory.

<b>Generator expression:</b> <b>(x * x for x in range(10))</b> — lazy, one value at a time.

<b>When to use which:</b> if you're passing the result to <b>sum()</b>, <b>max()</b>, <b>any()</b>, <b>all()</b>, or <b>join()</b> — a function that consumes the iterable once — use a generator expression. If you need the list later, use a list comprehension.

<h6>
<b> Agent-forward example — streaming </b>

When ChatGPT shows words appearing one at a time, the API is giving you a generator of tokens. The same shape: the server yields each token as it's generated, and the client consumes them in a loop.

This is why LLM APIs use generators: the response is not available all at once. Each token arrives milliseconds apart. If you waited for the full response, you'd stare at a blank screen for 10 seconds. With a generator, you see the first token in milliseconds and the rest stream in as they're ready.
</h6>

```python
def fake_llm_stream(text):
    for word in text.split():
        yield word + " "

for token in fake_llm_stream("Revenue grew 12 percent in Q4"):
    print(token, end="", flush=True)
```

<h6>
<b> Line-by-line </b>

`def fake_llm_stream(text):` — generator function. The `yield` inside makes it one.

`for word in text.split():` — iterate over the words in the string.

`yield word + " "` — produce one token (word plus a space) and pause.

`for token in fake_llm_stream(...):` — consumer. The for loop calls `next()` internally, gets each token, prints it, and waits for the next.

`end="", flush=True` — `end=""` prevents a newline after each token (so tokens appear on the same line). `flush=True` forces the terminal to display each token immediately rather than buffering.

<b>The shape matters:</b> the producer (the API) yields tokens one at a time. The consumer (your UI) receives them in a loop. Neither side needs to hold the full response in memory. In Week 9, your Streamlit chat UI displays tokens this way.
</h6>

<h6>
<b> Software-thinking example — processing a huge file </b>

Reading a 5 GB CSV with <b>f.read()</b> loads the entire file into memory. On most machines, the program crashes. A generator reads one row at a time. The file stays open between rows. Memory usage stays constant regardless of file size.

The <b>with</b> statement (context manager from the file I/O track) guarantees the file closes correctly even if the generator is interrupted mid-iteration.
</h6>

```python
import csv

def iter_rows(path):
    with open(path, newline="") as f:
        for row in csv.DictReader(f):
            yield row

missing = sum(1 for row in iter_rows("sales.csv") if not row["revenue"])
```

<h6>
<b> Line-by-line </b>

`def iter_rows(path):` — generator function.

`with open(path, newline="") as f:` — context manager. The file opens here and closes when the generator function finishes (or is garbage collected).

`for row in csv.DictReader(f):` — iterate over rows, one at a time.

`yield row` — hand back one row and pause. The file stays open. The position in the file is remembered.

`missing = sum(1 for row in iter_rows(...) if not row["revenue"])` — the generator expression <b>(1 for row in iter_rows(...) if ...)</b> produces a `1` for each row where the condition is true. `sum()` adds them up. This counts rows without ever materializing a list.

<b>Why it matters:</b> a 5 GB file, processed row by row, uses the memory of one row. The same file loaded with `f.read()` would use 5 GB. That's the difference between "works" and "crashes."
</h6>

<h6>
<b> What you should take away </b>

<b> 1. `yield` turns a function into a generator. </b>

A function with `yield` anywhere in its body is a generator function. Calling it doesn't run the body — it returns a generator object. The body runs when the consumer asks for values.

<b> 2. Generators are lazy — one value at a time, O(1) memory. </b>

Use a generator when you're producing a large sequence that the consumer will process once. Use a list when you need all values in memory at once, or when you need to iterate twice.

<b> 3. Generators can only be consumed once. </b>

After the last value, the generator is exhausted. Calling `next()` raises `StopIteration`. A `for` loop stops cleanly. If you iterate again, you get nothing.

<b> 4. Generator expressions are the one-line form. </b>

Use `(expr for item in iterable)` when you're passing the result to a consumer like `sum()`, `max()`, `any()`, `all()`, or `join()`. Use `[expr for item in iterable]` when you need the list itself.

<b> 5. LLM streaming is a generator. </b>

The ChatGPT API yields tokens one at a time. Your UI consumes them in a `for` loop. This is why streaming feels instant — the first token shows up in milliseconds, and the rest follow as they're generated.
