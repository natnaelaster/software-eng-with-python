<b> Concept 1: functions are objects (the prerequisite for decorators)</b>

<h6>
The analogy. A function is like a recipe card. You can read it, copy it, hand it to someone, or put it in a drawer next to other cards. 

Calling the function is cooking the recipe. Most beginners only ever cook. Today you start passing the card around.

This is a foundational concept. You've used functions. Today you learn to treat them like data — pass them, store them, return them.

<b> Part 1: The Core Idea</b>

A function in Python is an object. You can:

- Assign it to a variable

- Store it in a list or dict

- Pass it as an argument to another function

- Return it from a function

This is what makes decorators possible. Without it, none of the later concepts work.

<strong> Part 1: A function is a value </strong>

`def shout(text):` creates a function object and binds the name `shout` to it.

The name `shout` **is** the object. Adding parentheses (`shout()`) **calls** it. Without parentheses, you're holding the object itself.

That's the same as `x = 5` — `x` is a name bound to an int. `shout` is a name bound to a function. Both are values. You can reassign them, pass them, store them.

Three things you can do once functions are objects:

1. **Assign** to another name: `yell = shout`
2. **Store** in a data structure: `actions = {"loud": shout}`
3. **Pass** into another function: `apply_twice(shout, "hey")`

Each one is demonstrated below.

```python
def shout(text):
    return text.upper() + "!"

# 1. Assign to another name (no parentheses = no call)
yell = shout
print(shout("hello"))            # HELLO!

# 2. Store in a data structure
actions = {"loud": shout, "plain": str}
print(actions["loud"]("hi"))    # HI!

# 3. Pass INTO another function
def apply_twice(func, value):
    return func(func(value))

print(apply_twice(shout, "hey"))   # HEY!!
```

<h6>

<strong> Line-by-line </strong>

`yell = shout` — no parentheses on either side. `yell` is now a second name for the **same** function object. `yell is shout` is `True`.

`actions = {"loud": shout, "plain": str}` — functions stored as dict values. `actions["loud"]` returns the function; adding `("hi")` calls it.

`apply_twice(shout, "hey")` — `func` is bound to the `shout` function object. Inside, `func("hey")` → `"HEY!"`, then `func("HEY!")` → `"HEY!!"`.

Notice: `shout` was written **without** parentheses when passed to `apply_twice`. If you write `shout()`, Python calls it immediately and passes the **result** — not the function.

<strong> What about the expected output </strong>

The notes said `HEY!!!` (three exclamation marks). That was a typo in the original concept text — the actual output is `HEY!!` (two). This is why you **always run the code** and trust the actual output over the notes.

<strong> Part 2: Closures — a function that remembers </strong>

A function defined **inside** another function can use the outer function's variables, even after the outer function has finished.

This is called a **closure**: the inner function "closes over" the outer variables, capturing them for later use.

Why this works: when you call `make_multiplier(2)`, the local `factor` would normally be destroyed when the function returns. But because `multiply` still references it, Python keeps the variable alive — attached to the returned function.

Each call to `make_multiplier` creates a **new, independent** closure. `double` remembers `factor = 2`. `triple` remembers `factor = 3`. They do not share state.

</h6>

```python
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))   # 20
print(triple(10))   # 30
```

<h6>

<strong> Line-by-line </strong>

`def make_multiplier(factor):` — outer function, takes `factor`.

`def multiply(x):` — inner function, defined inside `make_multiplier`.

`return x * factor` — reads `factor` from the enclosing scope. This is what makes it a closure.

`return multiply` — returns the inner **function object** (no parentheses → no call). If you wrote `return multiply()`, Python would call it and try to return the result — which would fail because `multiply` needs an argument.

`double = make_multiplier(2)` — runs the outer function once. `factor = 2`. Returns the `multiply` function with `factor` captured. Now `double(10)` reads `factor = 2` → `20`.

`triple = make_multiplier(3)` — a **separate** call. Fresh `factor = 3`. Returns a **different** `multiply` function. `triple(10)` reads `factor = 3` → `30`.

Hint: look at whether `step` **modifies** `count` or only **reads** it. That distinction determines the output.

</h6>

```python
def make_counter():
    count = 0
    def step():
        return count + 1
    return step

c = make_counter()
print(c())
print(c())
```

<h6>

<strong> Prediction answer </strong>

Output:

1
1

<strong> Why </strong>

`step` only **reads** `count`. It never modifies it. There is no `count += 1`. So every call to `c()` reads the same `count = 0` and returns `0 + 1 = 1`.

The trap: beginners assume "counter" means "increments." But this code has no increment statement. The name `make_counter` is misleading — the function returns a value that never changes.

To make it actually count, you'd need `nonlocal count` and `count += 1`. That's covered in the "Try it yourself" exercise.

<strong> What you should take away </strong>

<b> 1. `shout` and `shout()` are different </b>

`shout` is the function object. `shout()` **calls** it. This distinction is the entire prerequisite for decorators — a decorator takes a function (no parentheses), returns a wrapped function (no parentheses).

<b> 2. Closures capture by reference, not by snapshot </b>

A closure sees the **current value** of the outer variable when it's called, not the value at the time it was defined. If the outer variable changes, the closure sees the change.

But a closure only sees changes if it **re-reads** the variable. And the inner function can only modify the outer variable with the `nonlocal` keyword.
