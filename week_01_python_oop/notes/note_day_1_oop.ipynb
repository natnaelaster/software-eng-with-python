**Concept 1: Class**

A class is a blueprint. An object is what you build from that blueprint. The blueprint for a house describes rooms, walls, and doors. Each actual house built from it is a separate object — same structure, different contents.

In Python, a class packages two things together: data (called attributes) and behavior (called methods). Once you define the blueprint, you can stamp out as many objects as you need, each with its own independent data.

example c1.1

```python
class dataprofile:
    def __init__(self, name, row, columns):
        self.name = name
        self.row = row
        self.columns = columns

    def summery(self):
        return f"{self.name}: {self.row} * {self.columns}"

    def area(self):
        return (self.row * self.columns)

    def is_large(self):
        return self.row > 100

sales = dataprofile('bet', 300, 40)
customers = dataprofile("foq", 220, 50)

print(sales.summery())
print(sales.is_large())
print(customers.summery())
print(sales.area())
print('+'*50)
```

**explanation**
- (__init__)runs automatically the moment you call DatasetProfile(...). It's the constructor. 
- (self) is just the object referring to itself — every instance method receives it as the first argument. 
- When you call sales.summary(), Python silently translates that to DatasetProfile.summary(sales).

**Concept 2 : instance attributes vs. class attributes**

**Class Attributes**
- Belong to the class itself, not to individual objects
- Shared across all instances of the class
- Defined outside the __init__ method
- Accessible via ClassName.attribute or instance.attribute
If modified via the class, affects all instances

**Instance Attributes**
- Belong to individual objects (instances)
- Unique to each instance
- Defined inside the __init__ method using self
- Accessible via instance.attribute
- Changes affect only that specific instance



example c2.1


```python
class DatasetProfile:
    # Class attribute — shared across ALL instances
    source_system = "internal_warehouse"
    instance_count = 0

    def __init__(self, name, rows, columns):
        # Instance attributes — unique per object
        self.name = name
        self.rows = rows
        self.columns = columns
        DatasetProfile.instance_count += 1   # mutate class attribute

    def summary(self):
        return (f"[{self.source_system}] "
                f"{self.name}: {self.rows}×{self.columns}")


sales = DatasetProfile("sales_q4", 250000, 18)
customers = DatasetProfile("customers", 42000, 9)

print(DatasetProfile.instance_count)   # 2
print(sales.source_system)             # internal_warehouse — found on class
sales.source_system = "external_api"   # NOW creates an instance attribute
print(sales.source_system)             # external_api — shadows class attr
print(customers.source_system)         # internal_warehouse — class attr unchanged
```

- The lookup order matters: Python looks at the instance first, then the class. When you assign sales.
- (source_system = "external_api"), you're creating a new instance attribute that shadows the class one — the class attribute itself is untouched.

**Concept 3: methods in depth**

Three kinds of methods exist. You need to know all three because they show up constantly in LangChain, FastAPI, and Pydantic.

example 3.1

```python
class DatasetProfile:
    source_system = "internal_warehouse"

    def __init__(self, name, rows, columns):
        self.name = name
        self.rows = rows
        self.columns = columns

    # INSTANCE method — operates on one object's data
    def memory_estimate_mb(self):
        bytes_per_cell = 8
        return (self.rows * self.columns * bytes_per_cell) / (1024 ** 2)

    # CLASS method — operates on the class itself, not an instance
    # Use when the method is about the class, not any particular object
    @classmethod
    def from_dataframe(cls, df, name):
        """Alternative constructor — build from a pandas DataFrame."""
        return cls(name, len(df), len(df.columns))

    # STATIC method — no access to instance or class
    # Just a utility that logically belongs here
    @staticmethod
    def valid_extension(filename):
        return filename.endswith((".csv", ".parquet", ".json"))


import pandas as pd

data = pd.DataFrame({
    "product": ["A", "B", "C"],
    "sales": [100, 200, 300],
    "profit": [10, 20, 30]
})

df = data  # your existing pandas workflow
profile = DatasetProfile.from_dataframe(df, "sales_q4")  # alternative constructor

print(DatasetProfile.valid_extension("data.csv"))     # True
print(DatasetProfile.valid_extension("report.xlsx"))  # False
```

- (@classmethod) is the pattern for alternative constructors — you'll see it everywhere in Python's standard library (datetime.fromisoformat, dict.fromkeys). 
- (@staticmethod) is for pure utility functions that happen to live on the class for organizational reasons.

```python
class dataprofile:
    source_sys = 'internal_warehouse'

    def __init__(self, name, row, column):
        self.name = name
        self.row = row
        self.column = column

    def estimate_mb(self):
        byte_per_cle =0.8
        return (self.row * self.column * byte_per_cle) / (1024 ** 2)

    @classmethod
    def from_dataframe(cls, df, name):
        return cls(name, len(df), len(df.columns))

    @staticmethod
    def valid_extension(filename):
        return filename.endswith(('.csv', '.parquet', '.json'))

data = pd.DataFrame({
    "product": ["A", "B", "C"], 
    "sales": [100, 200, 300],
    "profit": [10, 20, 30]
})

df = data  # your existing pandas workflow
profile = dataprofile.from_dataframe(df, "sales_q4")  # alternative constructor

print(dataprofile.valid_extension("data.csv"))     # True
print(dataprofile.valid_extension("report.xlsx"))  # False
print(profile)
```

```python
class modelresult:
    def __init__(self, name, accuracy, train_time):
        self.name = name
        self.accuracy = accuracy
        self.train_time = train_time

    def __repr__(self):
        return f'{self.name!r} = {self.accuracy}acc * {self.train_time}time'   

    def __str__(self):
        #byte_per_cell = 8
        return f'self.accuracy * self.train_time * byte_per_cell'

    def __eq__(self, other):
        return self.accuracy == other.accuracy 

    @property
    def accuracy(self):
        return self._accuracy

    def grade(self, value):
        if value >= 0.9:
            return 'A'
        elif value >= 0.8:
            return 'B'
        elif value >= 0.7:
            return 'C'
        else:
            return 'F'
        self._accuracy = value
    @accuracy.setter
    def accuracy(self, value):
        if 0 <= value <= 1:
            self._accuracy = value
        else:
            raise ValueError('accuracy must be between 0 & 1')
m1 = modelresult("RandomForest", 0.94, 12.3)
m2 = modelresult("RandomForest", 0.94, 45.1)
m3 = modelresult("XGBoost", 0.9, 12.3)

print(m1 == m2)  # should be True
print(m1 == m3)  # should be False
print(m1.grade(m1.accuracy))
m1.accuracy = 0.9
print(m1.accuracy)
print(repr(m3))
print(str(m3))
print(repr(m3))
```

Notice !r inside the f-string — that applies repr() to the value, which adds quotes around strings automatically. This is idiomatic Python.

**Concept 5: properties — controlled attribute access**

A @property lets you expose what looks like a plain attribute but runs logic underneath. This is how you add validation, computed values, or lazy loading without changing your API.

```python
class ColumnProfile:

    def __init__ (self, name: str, dtype: str, null_count: int, total_count: int):

        if null_count < 0:
            raise ValueError('null_count cannot be negative')
        if null_count > total_count:
            raise ValueError('null_count cannot be greater than total_count')    

        self.name = name
        self.dtype = dtype
        self.null_count = null_count
        self.total_count = total_count

    @property
    def null_rate(self) -> float:
        if self.total_count == 0:
            return 0
        return self.null_count / self.total_count

    @property
    def is_clean(self) -> bool:
        return self.null_rate < 0.05

    @classmethod
    def from_series(cls, series: pd.Series, name: str = None) -> 'ColumnProfile':
        if name is None:
            name = series.name if series.name is not None else 'unnamed'
        return cls(
            name = name,
            dtype = str(dtype),
            null_count = int(series.isna().sum()),
            total_count = len(series)
        )

    def __repr__(self) -> str:
        """Copy-pasteable string to reconstruct object."""
        return f"ColumnProfile(name='{self.name}', dtype='{self.dtype}', null_count={self.null_count}, total_count={self.total_count})"

        
    def __str__(self) -> str:
        status = "Clean" if self.is_clean else "Needs attention"
        return f"Column: {self.name} | Type: {self.dtype} | Nulls: {self.null_count}/{self.total_count} ({self.null_rate:.2%}) | Status: {status}"


```

```python
import pandas as pd

class ColumnProfile:

    def __init__(self, name: str, dtype: str, null_count: int, total_count:int):
        
        if null_count < 0:
            raise ValueError(f"null_count must be >= 0, got {null_count}")
        if null_count > total_count:
            raise ValueError(f"null_count ({null_count}) cannot exceed total_count ({total_count})")

        self.name = name
        self.dtype = dtype
        self.null_count = null_count
        self.total_count = total_count

    def __repr__(self) -> str:
        return (f"ColumnProfile(name={self.name!r}, dtype={self.dtype!r}, "
                f"null_count={self.null_count}, total_count={self.total_count})")

    def __str__(self) -> str:
        return f"Column '{self.name}' [{self.dtype}] — {self.null_rate:.1%} null"    

    @property
    def null_rate(self) -> float:
        if self.total_count == 0:
            return 0
        return self.null_count / self.total_count  

    @property
    def is_clean(self) -> bool:
        return self.null_rate < 0.05

    @classmethod
    def from_series(cls, series: pd.Series, name: str = None) -> 'ColumnProfile':
        if name is None: 
            name = series.name if series.name is not None else 'unnamed'
        return cls(
            name = name,
            dtype = str(series.dtype),
            null_count = int(series.isna().sum()),
            total_count = len(series)
        )       

class DatasetProfile:

    def __init__(self):
        self._columns: list[ColumnProfile] = []

    def add_column(self, col_profile: ColumnProfile) -> None:
        self._columns.append(col_profile)

    @property
    def clean_columns(self) -> list[ColumnProfile]:
        return [col for col in self._columns if col.is_clean]

    def quality_report(self) -> None:
        print('\n' + '='*50)
        print('Data Set Quality')
        print('+'*50)

        for col in self._columns:
            grade = 'clean' if col.is_clean else 'needs attention'
            print(f'{col.name:20} {col.null_rate:10.2%} {grade}')

        print('-'* 50)
        total = len(self._columns)  
        clean = len(self.clean_columns)    
        if total > 0:
            print(f"Total columns : {total}")
            print(f"Clean columns : {clean} ({clean / total * 100:.1f}%)")
        else:
            print("No columns profiled.")
        print('=' * 50 + '\n')     

    @classmethod
    def from_dataframe(cls, df: pd.DataFrame) -> 'DatasetProfile':
        profile = cls()
        for col_name in df.columns:
            profile.add_column(ColumnProfile.from_series(df[col_name]))
        return profile
```

```python
# Create sample data
df = pd.DataFrame({
    'name': ['Alice', 'Bob', None, 'David'],
    'age': [25, 30, 22, None],
    'score': [85.5, 92.3, 78.1, 89.7]
})

# Build profile
profile = DatasetProfile.from_dataframe(df)

# Print quality report
profile.quality_report()

# Access clean columns
print("Clean columns:", [col.name for col in profile.clean_columns])

# Create single column profile manually
col = ColumnProfile('age', 'int64', 1, 4)
print(col)
print(repr(col))
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2
print(v3.x, v3.y)
```

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    
    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32

t = Temperature(25)
print(t.fahrenheit)
```

```python
def my_decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```
