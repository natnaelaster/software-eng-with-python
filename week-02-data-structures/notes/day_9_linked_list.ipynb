<strong><i> **Concept 1: a chain of nodes, each pointing to the next**</i> </strong>
<h6>

<strong> The general idea </strong>

A linked list is a way of storing a sequence of items where each item lives in its own object, called a <b>node</b>, and each node holds a reference (a pointer) to the next node. The list itself is just a reference to the <b>first</b> node — the <b>head</b>. From the head, you follow the chain of references to reach every other node.

This is different from an array. An array stores all elements in one contiguous block, so you can jump to any index with `start + i * slot_size`. A linked list has no contiguous block. The nodes are scattered in memory, connected only by references. To reach the k-th node, you must start at the head and follow k-1 references. That's O(k), not O(1).

The trade-off is the whole point of linked lists. You lose O(1) random access. But you gain O(1) insertion and deletion at the head. In an array, inserting at the front is O(n) because every element must shift right. In a linked list, inserting at the front is O(1) — just change one reference. That's why queues (which always remove from the front) are often built on linked lists.
</h6>

<h6>
<strong> Example 1: building a chain by hand </strong>

Three nodes, connected by references. The list starts at <b>a</b>. Following <b>a.next</b> gives <b>b</b>. Following <b>b.next</b> gives <b>c</b>. Following <b>c.next</b> gives <b>None</b> — the end of the chain.
</h6>



````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

a, b, c = Node(1), Node(2), Node(3)
a.next = b
b.next = c

current = a
while current is not None:
    print(current.value)
    current = current.next
````

<h6>
<strong> Line-by-line </strong>

<b>class Node:</b> — a node holds a value and a reference to the next node.

<b>self.value = value</b> — the data the node carries.

<b>self.next = None</b> — by default, no next node. The chain ends here.

<b>a, b, c = Node(1), Node(2), Node(3)</b> — three separate nodes, each with its own value. Initially, none of them point to each other.

<b>a.next = b</b> — node <b>a</b>'s reference now points to <b>b</b>. The chain starts forming.

<b>b.next = c</b> — node <b>b</b> now points to <b>c</b>. The chain is: <b>a → b → c → None</b>.

<b>current = a</b> — start at the head.

<b>while current is not None:</b> — walk until the chain ends.

<b>print(current.value)</b> — print the node's value.

<b>current = current.next</b> — follow the reference to the next node.

<b>The output:</b> 1, then 2, then 3. Then <b>current</b> becomes <b>None</b>, and the loop ends.
</h6>

<h6>
<strong> Example 2: inserting at the head (O(1)) </strong>

Inserting a new node at the head is the fastest operation in a linked list. Create a node, point it at the current head, and update the head reference. Three operations. No shifting. O(1).
</h6>

````python
def insert_at_head(head, value):
    new_node = Node(value)
    new_node.next = head
    return new_node

head = Node(2)
head = insert_at_head(head, 1)   # 1 → 2
head = insert_at_head(head, 0)   # 0 → 1 → 2
````

<h6>
<strong> Line-by-line </strong>

<b>def insert_at_head(head, value):</b> — takes the current head and a value. Returns the new head.

<b>new_node = Node(value)</b> — create a new node.

<b>new_node.next = head</b> — the new node points to the old head.

<b>return new_node</b> — the new node is now the head.

<b>head = insert_at_head(head, 1)</b> — the head is now the new node (value 1), which points to the old head (value 2).

<b>head = insert_at_head(head, 0)</b> — the head is now 0, pointing to 1, pointing to 2.

<b>Why this is O(1):</b> three operations — create, point, return. No loop over the existing nodes. In an array, inserting at the front requires shifting every element right, which is O(n). That difference is the entire reason linked lists exist.
</h6>

<h6>
<strong> Example 3: deleting from the head (O(1)) </strong>

Deleting the head is equally cheap. Just update the head reference to the next node. The old head is garbage-collected (Python frees it automatically).
</h6>

````python
def delete_head(head):
    if head is None:
        return None
    return head.next
````

<h6>
<strong> Line-by-line </strong>

<b>if head is None:</b> — the list is empty. Nothing to delete. Return <b>None</b>.

<b>return head.next</b> — the new head is the old head's next node. One operation. O(1).

<b>Why this matters for queues:</b> a queue removes from the front. In an array, removing the front is O(n) — every element shifts left. In a linked list, removing the front is O(1). That's why queues are often built on linked lists.
</h6>

<h6>
<strong> Example 4: searching (O(n)) </strong>

Reaching the k-th node is O(k). Finding a value is O(n) — you might have to walk the whole chain. There's no address arithmetic, so you can't jump. You start at the head and follow references.
</h6>

````python
def length(head):
    count = 0
    current = head
    while current is not None:
        count += 1
        current = current.next
    return count

def contains(head, value):
    current = head
    while current is not None:
        if current.value == value:
            return True
        current = current.next
    return False
````

<h6>
<strong> Line-by-line </strong>

<b>def length(head):</b> — count the nodes.

<b>count = 0</b> — start at zero.

<b>current = head</b> — start at the head.

<b>while current is not None:</b> — walk until the chain ends.

<b>count += 1</b> — one more node.

<b>current = current.next</b> — follow the reference.

<b>return count</b> — the total.

<b>def contains(head, value):</b> — search for a value.

<b>current = head</b> — start at the head.

<b>while current is not None:</b> — walk the chain.

<b>if current.value == value:</b> — found it.

<b>return True</b> — exit early.

<b>current = current.next</b> — keep walking.

<b>return False</b> — never found it.

<b>Why both are O(n):</b> in the worst case, you visit every node. There's no shortcut. That's the cost of losing contiguous storage.
</h6>

<h6>
<strong> Example 5: doubly linked list </strong>

A doubly linked list adds a <b>prev</b> reference to each node, so you can walk backward as well as forward. This makes deletion easier (you don't need to track the previous node separately), and it enables O(1) deletion if you have a reference to the node.
</h6>

````python
class DoubleNode:
    def __init__(self, value):
        self.value = value
        self.next = None
        self.prev = None
````

<h6>
<strong> Line-by-line </strong>

<b>self.next = None</b> — forward reference, same as before.

<b>self.prev = None</b> — new: backward reference.

<b>Why doubly linked lists exist:</b> in a singly linked list, to delete a node, you need a reference to the node <b>before</b> it. If you only have the node to delete, you must walk from the head to find its predecessor — that's O(n). In a doubly linked list, the node has a <b>prev</b> reference, so you can delete it in O(1).

<b>The cost:</b> each node stores two references instead of one. More memory per node, but faster deletion.
</h6>

<h6>
<strong> Example 6: circular linked list </strong>

In a circular linked list, the last node points back to the head. There is no <b>None</b> at the end. Walking past the last node brings you back to the first. This is useful for round-robin scheduling (like a music playlist on repeat).
</h6>

````python
# Circular: c.next = a instead of None
a, b, c = Node(1), Node(2), Node(3)
a.next = b
b.next = c
c.next = a   # circular!

# Walk exactly 6 steps to avoid infinite loop
current = a
for _ in range(6):
    print(current.value)
    current = current.next
````

<h6>
<strong> Line-by-line </strong>

<b>c.next = a</b> — instead of <b>None</b>, the last node points back to the head. The chain is now a circle.

<b>for _ in range(6):</b> — walk a fixed number of steps. <b>while current is not None</b> would loop forever, because there's no <b>None</b>.

<b>print(current.value)</b> — output: 1, 2, 3, 1, 2, 3.

<b>current = current.next</b> — after 3, the next is 1 again.

<b>Why circular lists exist:</b> round-robin scheduling, buffering, and any problem where "after the last item, start over." The <b>None</b> sentinel is replaced by the head, and the traversal loop needs a step counter instead of a <b>None</b> check.
</h6>

<h6>
<strong> From your note </strong>

Your note uses the treasure hunt analogy: each clue card holds a value and tells you where the next card is. You can't jump to card 5 — you start at card 1 and follow directions. The last card says "no more clues."

That's the linked list in one paragraph. Each node is a clue card. The value is the treasure. The <b>next</b> reference is the direction to the next card. <b>None</b> means "no more clues."

It's just orphaned from the chain.

<b>Why:</b>

<b>The key insight:</b> a linked list is defined by <b>reachability from the head</b>, not by the existence of the nodes. Any node not reachable from the head is effectively gone.

<strong> What you should take away </strong>

<b> 1. A linked list is a chain of nodes connected by references. </b>

Each node holds a value and a reference to the next. The list is the head plus everything reachable by following references. There is no contiguous block, no address arithmetic.

<b> 2. O(1) insert and delete at the head. </b>

This is the main advantage over arrays. Queues (which always remove from the front) are often built on linked lists because of this.

<b> 3. O(n) access and search. </b>

Reaching the k-th node is O(k). Searching for a value is O(n). You lose O(1) random access — that's the trade-off.

<b> 4. Doubly linked lists add a prev reference. </b>

O(1) deletion if you have the node reference. Costs more memory per node.

<b> 5. Circular linked lists have no None at the end. </b>

The last node points back to the head. Useful for round-robin scheduling. Traversal needs a step counter, not a None check.
</h6>
