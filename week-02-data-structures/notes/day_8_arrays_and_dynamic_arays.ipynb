<h6>
<strong><u> Concept 1: an array is a row of equal-size slots </u></strong>


<strong> The analogy </strong>

A parking garage where every space is exactly 8 feet wide, and space 0 starts at the entrance. To find space 5, you don't walk past spaces 0 to 4 and count. You multiply: 5 × 8 feet from the entrance. That one multiplication is why an array can jump to any element instantly.
</h6>

<h6>
An array is a **contiguous block of equal-size slots**. Every slot occupies exactly the same number of bytes. Slot 0 starts at some base address. Slot 1 starts exactly `slot_size` bytes after slot 0. Slot 2 starts exactly `slot_size` bytes after slot 1. And so on.

Because every slot is the same width and they sit side by side, the address of slot `i` is:

**address = start + i × slot_size**

One multiplication, one addition. That's it. No search, no loop, no comparison. The same two operations find slot 5 and slot 1,000,000. **That's what O(1) means at the hardware level** — the work doesn't grow with the index.
</h6>

````python
def slot_address(start: int, index: int, slot_size: int) -> int:
    """Where an array element lives in memory: no searching, just arithmetic."""
    return start + index * slot_size

print(slot_address(1000, 0, 8))           # 1000
print(slot_address(1000, 5, 8))           # 1040
print(slot_address(1000, 1_000_000, 8))   # 8001000: still one multiplication
````

<h6>
<strong> Line-by-line </strong>

`def slot_address(start, index, slot_size):` — three integers in, one integer out. `start` is the address of slot 0. `index` is the slot number. `slot_size` is the width of each slot in bytes.

`return start + index * slot_size` — the entire formula. One multiplication (`index * slot_size`) and one addition (`+ start`). No loop.

`slot_address(1000, 0, 8)` → `1000 + 0 * 8 = 1000`. Slot 0 lives at the start address.

`slot_address(1000, 5, 8)` → `1000 + 5 * 8 = 1000 + 40 = 1040`. Slot 5 lives 40 bytes after the start.

`slot_address(1000, 1_000_000, 8)` → `1000 + 1_000_000 * 8 = 1000 + 8_000_000 = 8_001_000`. Slot 1,000,000 lives 8,000,000 bytes after the start. **Still one multiplication.**

The function does the same work for index 5 and index 1,000,000. That's the definition of **O(1)** — constant time regardless of input size. It works only because the slots are equal-size and contiguous.
</h6>

<h6>
<b> Why this matters — the real reason arrays are fast </b>

Beginners often think "arrays are fast because the CPU is fast." That's wrong. Arrays are fast because **the address formula doesn't depend on the index**. Whether you're reading `a[0]` or `a[999999]`, the CPU does exactly two operations: multiply, add.

A linked list, by contrast, cannot do this. To find the 1,000,000th node, you must walk through 999,999 nodes. Each step is a memory read. That's O(n), because the work grows with the index.

**The equal-size + contiguous layout is the entire reason arrays get O(1) random access.** Break either condition — variable-size elements, or non-contiguous storage — and O(1) access is lost.

This is the foundation of every array-like structure you'll build in Week 2: dynamic arrays, hash tables (which use arrays internally), and heaps (which are arrays with a tree shape imposed on them).
</h6>

<b> What you should take away </b>

The address of slot `i` is `start + i * slot_size`. Same two operations for every index. That's why access time doesn't grow with the index. It's a formula, not a search.

<b> Challenge 1: Running Sum of 1d Array</b>

**Source:** LeetCode 1480 (Easy)

**Problem:** Given an array `nums`, return an array where each element at index `i` is the sum of `nums[0]` through `nums[i]`.

Example: `[1, 2, 3, 4]` → `[1, 3, 6, 10]`.

**What to practice:** O(n) single-pass accumulation. Reading `nums[i]` and writing `result[i]` — both O(1) per access, so total O(n).

**Starter:**

```python
def running_sum(nums):
    # your code here
    pass

# test
print(running_sum([1, 2, 3, 4]))   # expected: [1, 3, 6, 10]
print(running_sum([1, 1, 1, 1]))   # expected: [1, 2, 3, 4]

````python
def running_sum(nums: list[int]) -> list[int]:
    for i in range(1, len(nums)):
        nums[i] += nums[i - 1]
    return nums

if __name__ == "__main__":
    print(running_sum([1, 2, 3, 4]))
````

<h6>
<b> Challenge 2: Find Pivot Index</b>

**Source:** LeetCode 724 (Easy)

**Problem:** Given an array `nums`, return the leftmost pivot index — the index where the sum of the numbers strictly to the left equals the sum of the numbers strictly to the right. If no such index exists, return -1.

Example: `[1, 7, 3, 6, 5, 6]` → `3` (because left of index 3 is `1+7+3=11`, right is `5+6=11`).

**What to practice:** prefix/suffix sums. Build a running total in one pass, then check each index's left sum against `total - left - nums[i]`. Every access to `nums[i]` is O(1); the whole solution is O(n).

**Starter:**

```python
def pivot_index(nums):
    # your code here
    pass

# test
print(pivot_index([1, 7, 3, 6, 5, 6]))   # expected: 3
print(pivot_index([1, 2, 3]))             # expected: -1
print(pivot_index([2, 1, -1]))            # expected: 0

````python
def pivot_index(nums: list[int]) -> int:
    total_sum = sum(nums)
    left_sum = 0
    
    for i, num in enumerate(nums):
        # Right sum is the total minus everything to the left and the current element
        if left_sum == total_sum - left_sum - num:
            return i
        left_sum += num
        
    return -1

if __name__ == "__main__":
    # Example usage:
    print(pivot_index([1, 7, 3, 6, 5, 6]))  # Output: 3
    print(pivot_index([1, 2, 3]))           # Output: -1
    print(pivot_index([2, 1, -1]))          # Output: 0
````

<strong> **Concept 2: a Python list is an array of references**</strong>

<h6>
<strong> The general idea </strong>

A Python list is not a collection of objects. It is a <b>collection of pointers</b> — each slot in the list holds a reference to an object that lives somewhere else in memory. The list itself is just a row of identical-sized slots. The objects can be any size, any type, anywhere in memory. The list doesn't care.

This is why you can put an integer, a string, and a float in the same list. In C or Java, an array of ints can only hold ints. In Python, a list of "anything" holds references to anything, and the references themselves are all the same size.

When you write <b>a = [1, 2, 3]</b>, you're creating a list object with three slots, each holding a pointer. When you write <b>b = a</b>, you copy the pointer from <b>a</b> to <b>b</b>. You do <b>not</b> copy the list. Both names now point to the same list object. 
Mutate the list through one name, and the other name sees the change.

The same thing happens when you multiply a list: <b>[x] * 3</b> creates a list with three slots, but <b>each slot holds the same reference to x</b>. If <b>x</b> is immutable (an int, a string), this is invisible — you can't mutate an int. If <b>x</b> is mutable (a list, a dict), every slot points to the same object. Change it through one slot, every slot sees the change.

This is the single most common source of confusing bugs in Python code. It's also the reason that building a 2D grid with <b>[[0] * 3] * 3</b> produces a bug that only shows up on the <b>second</b> write — the first write works, which is the worst kind of bug.
</h6>

<h6>
<strong> Example 1: lists hold references, not values </strong>

The list holds three references. One points to an int, one to a string, one to a float. The slots are all the same size (one pointer each). This is why mixed-type lists work.
</h6>

````python
items = [10, "agent", 3.5]
print([type(x).__name__ for x in items])    # ['int', 'str', 'float']
````


<h6>
<strong> Example 2: assignment copies the reference </strong>

<b>b = a</b> copies the pointer, not the list. Both names point to the same object. <b>is</b> checks identity — whether two names point to the same object. If <b>a is b</b> is <b>True</b>, mutating one mutates the other.
</h6>

````python
a = [1, 2]
b = a
b.append(3)
print(a)        # [1, 2, 3]
print(a is b)   # True
````

<h6>
<strong> Example 3: multiplication copies references </strong>

<b>[[0] * 3] * 3</b> creates <b>one</b> inner list, then copies its reference three times. All three "rows" are the same list. Setting <b>grid[0][0]</b> changes all three, because there's only one inner list.
</h6>

````python
grid = [[0] * 3] * 3
grid[0][0] = 7
print(grid)     # [[7, 0, 0], [7, 0, 0], [7, 0, 0]]
````

<h6>
<strong> Example 4: the fix — build a new list each time </strong>

A comprehension runs the expression <b>once per iteration</b>, creating a <b>new</b> inner list each time. Three independent rows. Only the first row changes.
</h6>

````python
grid = [[0] * 3 for _ in range(3)]
grid[0][0] = 7
print(grid)                 # [[7, 0, 0], [0, 0, 0], [0, 0, 0]]
print(grid[0] is grid[1])   # False
````

<h6>
<strong> Example 5: the same bug in adjacency lists </strong>

Every graph problem that stores a graph as a list of lists runs into this. The fix is always a comprehension.
</h6>

````python
# Wrong
graph = [[]] * 4
graph[0].append(1)
print(graph)    # [[1], [1], [1], [1]]

# Right
graph = [[] for _ in range(4)]
graph[0].append(1)
print(graph)    # [[1], [], [], []]
````

<h6>
<strong> Example 6: shallow copy vs. deep copy </strong>

<b>.copy()</b> creates a new outer list, but the inner lists are shared. <b>copy.deepcopy()</b> recursively copies everything. Use <b>deepcopy</b> when you'll mutate nested objects.
</h6>

````python
import copy

matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
shallow[0][0] = 99
print(matrix)                  # [[99, 2], [3, 4]]

matrix = [[1, 2], [3, 4]]
deep = copy.deepcopy(matrix)
deep[0][0] = 99
print(matrix)                  # [[1, 2], [3, 4]]
````

<h6>
<strong> From the note </strong>

Your note says: <i>"A Python list is an array of references. Every slot in a Python list is the same size (one reference), so index lookup is still O(1). The objects they point to can be any size and any type."</i>

That's the mechanism. The consequence is that <b>assignment and multiplication copy references, not objects</b>. The trap is that mutable objects (lists, dicts, sets) share state. The fix is a comprehension or a <b>.copy()</b>.

The prediction question — <b>[[0] * 3] * 3</b> — demonstrates the trap. The answer is <b>[[7, 0, 0], [7, 0, 0], [7, 0, 0]]</b> because all three rows are the same object.
</h6>

<strong> What you should take away </strong>

<b> 1. Python lists hold references, not values. </b>

Assignment and multiplication copy references. Mutating a shared mutable object affects every name that points to it.

<b> 2. `[x] * n` is a trap for mutable `x`. </b>

For ints and strings, it's safe. For lists, dicts, and sets, it shares the same object across all slots.

<b> 3. Fix the trap with a comprehension or `.copy()`. </b>

<b>[[0] * n for _ in range(m)]</b> creates independent rows. <b>.copy()</b> creates a new outer list, but nested objects are still shared. <b>copy.deepcopy()</b> is fully independent.

<b> 4. This bug appears in grids, buckets, adjacency lists, memo tables, and backtracking. </b>

Whenever you build a container of mutable containers, ask: "Is each inner container a new object?" Use <b>is</b> to check.
