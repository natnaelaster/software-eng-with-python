<strong>Concept 1: an array is a row of equal-size slots </strong>


<h6>
<strong> The analogy </strong>

A parking garage where every space is exactly 8 feet wide, and space 0 starts at the entrance. To find space 5, you don't walk past spaces 0 to 4 and count. You multiply: 5 × 8 feet from the entrance. That one multiplication is why an array can jump to any element instantly.
</h6>

<h6>
An array is a **contiguous block of equal-size slots**. Every slot occupies exactly the same number of bytes. Slot 0 starts at some base address. Slot 1 starts exactly `slot_size` bytes after slot 0. Slot 2 starts exactly `slot_size` bytes after slot 1. And so on.

Because every slot is the same width and they sit side by side, the address of slot `i` is:

**address = start + i × slot_size**

One multiplication, one addition. That's it. No search, no loop, no comparison. The same two operations find slot 5 and slot 1,000,000. **That's what O(1) means at the hardware level** — the work doesn't grow with the index.
</h6>

````python
def slot_address(start: int, index: int, slot_size: int) -> int:
    """Where an array element lives in memory: no searching, just arithmetic."""
    return start + index * slot_size

print(slot_address(1000, 0, 8))           # 1000
print(slot_address(1000, 5, 8))           # 1040
print(slot_address(1000, 1_000_000, 8))   # 8001000: still one multiplication
````

<h6>
<strong> Line-by-line </strong>

`def slot_address(start, index, slot_size):` — three integers in, one integer out. `start` is the address of slot 0. `index` is the slot number. `slot_size` is the width of each slot in bytes.

`return start + index * slot_size` — the entire formula. One multiplication (`index * slot_size`) and one addition (`+ start`). No loop.

`slot_address(1000, 0, 8)` → `1000 + 0 * 8 = 1000`. Slot 0 lives at the start address.

`slot_address(1000, 5, 8)` → `1000 + 5 * 8 = 1000 + 40 = 1040`. Slot 5 lives 40 bytes after the start.

`slot_address(1000, 1_000_000, 8)` → `1000 + 1_000_000 * 8 = 1000 + 8_000_000 = 8_001_000`. Slot 1,000,000 lives 8,000,000 bytes after the start. **Still one multiplication.**

The function does the same work for index 5 and index 1,000,000. That's the definition of **O(1)** — constant time regardless of input size. It works only because the slots are equal-size and contiguous.
</h6>

<h6>
<b> Why this matters — the real reason arrays are fast </b>

Beginners often think "arrays are fast because the CPU is fast." That's wrong. Arrays are fast because **the address formula doesn't depend on the index**. Whether you're reading `a[0]` or `a[999999]`, the CPU does exactly two operations: multiply, add.

A linked list, by contrast, cannot do this. To find the 1,000,000th node, you must walk through 999,999 nodes. Each step is a memory read. That's O(n), because the work grows with the index.

**The equal-size + contiguous layout is the entire reason arrays get O(1) random access.** Break either condition — variable-size elements, or non-contiguous storage — and O(1) access is lost.

This is the foundation of every array-like structure you'll build in Week 2: dynamic arrays, hash tables (which use arrays internally), and heaps (which are arrays with a tree shape imposed on them).
</h6>

<b> Challenge 1: Running Sum of 1d Array</b>

**Source:** LeetCode 1480 (Easy)

**Problem:** Given an array `nums`, return an array where each element at index `i` is the sum of `nums[0]` through `nums[i]`.

Example: `[1, 2, 3, 4]` → `[1, 3, 6, 10]`.

**What to practice:** O(n) single-pass accumulation. Reading `nums[i]` and writing `result[i]` — both O(1) per access, so total O(n).

**Starter:**

```python
def running_sum(nums):
    # your code here
    pass

# test
print(running_sum([1, 2, 3, 4]))   # expected: [1, 3, 6, 10]
print(running_sum([1, 1, 1, 1]))   # expected: [1, 2, 3, 4]

````python
def running_sum(nums: list[int]) -> list[int]:
    for i in range(1, len(nums)):
        nums[i] += nums[i - 1]
    return nums

if __name__ == "__main__":
    print(running_sum([1, 2, 3, 4]))
````

<h6>
<b> Challenge 2: Find Pivot Index</b>

**Source:** LeetCode 724 (Easy)

**Problem:** Given an array `nums`, return the leftmost pivot index — the index where the sum of the numbers strictly to the left equals the sum of the numbers strictly to the right. If no such index exists, return -1.

Example: `[1, 7, 3, 6, 5, 6]` → `3` (because left of index 3 is `1+7+3=11`, right is `5+6=11`).

**What to practice:** prefix/suffix sums. Build a running total in one pass, then check each index's left sum against `total - left - nums[i]`. Every access to `nums[i]` is O(1); the whole solution is O(n).

**Starter:**

```python
def pivot_index(nums):
    # your code here
    pass

# test
print(pivot_index([1, 7, 3, 6, 5, 6]))   # expected: 3
print(pivot_index([1, 2, 3]))             # expected: -1
print(pivot_index([2, 1, -1]))            # expected: 0

````python
def pivot_index(nums: list[int]) -> int:
    total_sum = sum(nums)
    left_sum = 0
    
    for i, num in enumerate(nums):
        # Right sum is the total minus everything to the left and the current element
        if left_sum == total_sum - left_sum - num:
            return i
        left_sum += num
        
    return -1

if __name__ == "__main__":
    # Example usage:
    print(pivot_index([1, 7, 3, 6, 5, 6]))  # Output: 3
    print(pivot_index([1, 2, 3]))           # Output: -1
    print(pivot_index([2, 1, -1]))          # Output: 0
````

<strong> **Concept 2: a Python list is an array of references**</strong>

---

<h6>
<strong> The general idea </strong>

A Python list is not a collection of objects. It is a <b>collection of pointers</b> — each slot in the list holds a reference to an object that lives somewhere else in memory. The list itself is just a row of identical-sized slots. The objects can be any size, any type, anywhere in memory. The list doesn't care.

This is why you can put an integer, a string, and a float in the same list. In C or Java, an array of ints can only hold ints. In Python, a list of "anything" holds references to anything, and the references themselves are all the same size.

When you write <b>a = [1, 2, 3]</b>, you're creating a list object with three slots, each holding a pointer. When you write <b>b = a</b>, you copy the pointer from <b>a</b> to <b>b</b>. You do <b>not</b> copy the list. Both names now point to the same list object. 
Mutate the list through one name, and the other name sees the change.

The same thing happens when you multiply a list: <b>[x] * 3</b> creates a list with three slots, but <b>each slot holds the same reference to x</b>. If <b>x</b> is immutable (an int, a string), this is invisible — you can't mutate an int. If <b>x</b> is mutable (a list, a dict), every slot points to the same object. Change it through one slot, every slot sees the change.

This is the single most common source of confusing bugs in Python code. It's also the reason that building a 2D grid with <b>[[0] * 3] * 3</b> produces a bug that only shows up on the <b>second</b> write — the first write works, which is the worst kind of bug.
</h6>

<h6>
<strong> Example 1: lists hold references, not values </strong>

The list holds three references. One points to an int, one to a string, one to a float. The slots are all the same size (one pointer each). This is why mixed-type lists work.
</h6>

````python
items = [10, "agent", 3.5]
print([type(x).__name__ for x in items])    # ['int', 'str', 'float']
````


<h6>
<strong> Example 2: assignment copies the reference </strong>

<b>b = a</b> copies the pointer, not the list. Both names point to the same object. <b>is</b> checks identity — whether two names point to the same object. If <b>a is b</b> is <b>True</b>, mutating one mutates the other.
</h6>

````python
a = [1, 2]
b = a
b.append(3)
print(a)        # [1, 2, 3]
print(a is b)   # True
````

<h6>
<strong> Example 3: multiplication copies references </strong>

<b>[[0] * 3] * 3</b> creates <b>one</b> inner list, then copies its reference three times. All three "rows" are the same list. Setting <b>grid[0][0]</b> changes all three, because there's only one inner list.
</h6>

````python
grid = [[0] * 3] * 3
grid[0][0] = 7
print(grid)     # [[7, 0, 0], [7, 0, 0], [7, 0, 0]]
````

<h6>
<strong> Example 4: the fix — build a new list each time </strong>

A comprehension runs the expression <b>once per iteration</b>, creating a <b>new</b> inner list each time. Three independent rows. Only the first row changes.
</h6>

````python
grid = [[0] * 3 for _ in range(3)]
grid[0][0] = 7
print(grid)                 # [[7, 0, 0], [0, 0, 0], [0, 0, 0]]
print(grid[0] is grid[1])   # False
````

<h6>
<strong> Example 5: the same bug in adjacency lists </strong>

Every graph problem that stores a graph as a list of lists runs into this. The fix is always a comprehension.
</h6>

````python
# Wrong
graph = [[]] * 4
graph[0].append(1)
print(graph)    # [[1], [1], [1], [1]]

# Right
graph = [[] for _ in range(4)]
graph[0].append(1)
print(graph)    # [[1], [], [], []]
````

<h6>
<strong> Example 6: shallow copy vs. deep copy </strong>

<b>.copy()</b> creates a new outer list, but the inner lists are shared. <b>copy.deepcopy()</b> recursively copies everything. Use <b>deepcopy</b> when you'll mutate nested objects.
</h6>

````python
import copy

matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
shallow[0][0] = 99
print(matrix)                  # [[99, 2], [3, 4]]

matrix = [[1, 2], [3, 4]]
deep = copy.deepcopy(matrix)
deep[0][0] = 99
print(matrix)                  # [[1, 2], [3, 4]]
````

<h6>
<strong> From the note </strong>

Your note says: <i>"A Python list is an array of references. Every slot in a Python list is the same size (one reference), so index lookup is still O(1). The objects they point to can be any size and any type."</i>

That's the mechanism. The consequence is that <b>assignment and multiplication copy references, not objects</b>. The trap is that mutable objects (lists, dicts, sets) share state. The fix is a comprehension or a <b>.copy()</b>.

The prediction question — <b>[[0] * 3] * 3</b> — demonstrates the trap. The answer is <b>[[7, 0, 0], [7, 0, 0], [7, 0, 0]]</b> because all three rows are the same object.
</h6>

<strong> **Concept 3: dynamic arrays and over-allocation**</strong>

---

<h6>
<strong> The general idea </strong>

Arrays in C or Java have a fixed size. When you create an array of 10 ints, it holds exactly 10 ints. You can't make it hold 11. To grow, you must create a new array of 20, copy the 10 elements over, and delete the old one. That's expensive — O(n) for a single append.

But Python lists are different. They don't have a fixed size. You can append and append and append, and the list keeps growing. How? The answer is that Python lists are <b>dynamic arrays</b>: behind the scenes, they over-allocate space. When the list fills up, Python allocates a bigger block (usually twice the size), copies the references over, and discards the old block. That copy is O(n), but it happens <b>rarely</b> — roughly once per doubling. So the cost is <b>amortized</b>: it's expensive on the resize, but the vast majority of appends are O(1).

The math: if the list doubles every time it fills, then over n appends, the total work is about 2n copies (summing 1 + 2 + 4 + ... + n). That's O(n) total, or <b>O(1) amortized</b> per append. The resize is expensive, but it's rare enough that it doesn't dominate.

This concept matters because it's the foundation of every "growable" data structure: Python lists, Python dicts, Java's ArrayList, C++'s std::vector, Go's slices. They all over-allocate. They all amortize.
</h6>

<h6>
<strong> Example 1: watching the resizes happen </strong>

The list's size in bytes jumps in steps. Most appends don't change the size — the list has spare space. When the spare space runs out, the size jumps. The resizes get rarer as the list grows.
</h6>

````python
import sys

nums = []
last = sys.getsizeof(nums)                   # size of the list object in bytes
print(f"len=0  bytes={last}")
print(f'-'* 20)
for i in range(1, 21):
    nums.append(i)
    size = sys.getsizeof(nums)
    if size != last:                         # size changed, so a resize just happened
        print(f"len={i:<2} bytes={size}")
        last = size
````

<h6>
<strong> Line-by-line </strong>

<b>nums = []</b> — an empty list. It still has some overhead: the object header, the reference to its internal array, its length, its capacity. <b>sys.getsizeof</b> reports this.

<b>last = sys.getsizeof(nums)</b> — record the starting size.

<b>for i in range(1, 21):</b> — 20 appends.

<b>nums.append(i)</b> — add one element. Most of the time this is O(1) — the list has spare capacity, so it just fills the next slot.

<b>size = sys.getsizeof(nums)</b> — check the size again.

<b>if size != last:</b> — if the size changed, a resize just happened. Print the new length and size.

Typical output (exact bytes vary by Python version):

```
len=0 bytes=56
len=1 bytes=88
len=5 bytes=120
len=9 bytes=184
len=17 bytes=248
```


Twenty appends, but only five size changes. The gaps:

- 0 → 1: allocate space for 1
- 1 → 5: allocate space for 4 (spare: 3)
- 5 → 9: allocate space for 8 (spare: 3)
- 9 → 17: allocate space for 16 (spare: 7)

The spare capacity grows. So resizes get <b>rarer</b> as the list grows, even though the list is getting bigger. That's the over-allocation paying off.
</h6>

<h6>
<strong> Example 2: amortized cost — why appends feel free </strong>

A single append is O(1) if there's spare capacity. If not, it's O(n) — but that only happens once every doubling. So over n appends, the total cost is O(n), which is <b>O(1) amortized</b>.

This is the whole reason Python lists are practical. If appends were O(n) every time, building a list of 1 million items would take O(n²) = 10¹² operations — hours. With amortization, it's O(n) = 10⁶ — seconds.
</h6>

````python
# The math behind amortization:

# Append 1: no resize (capacity was 0 → grow to 1, cost 1)
# Append 2: no resize (capacity 1 → grow to 2, cost 2)
# Append 3: no resize
# Append 4: no resize (capacity 3 → grow to 4, cost 4)
# ...
# Total work for n appends: 1 + 2 + 4 + 8 + ... + n = 2n
# That's O(n) total, so O(1) per append on average.
````

<h6>
<strong> Example 3: the cost of inserting in the middle </strong>

Appending is amortized O(1). But inserting in the middle is O(n) — every element after the insertion point must shift right. That's not amortized; it's expensive <b>every time</b>. This is why "append, don't insert" is a rule of thumb for list-heavy code.
</h6>

````python
nums = [10, 20, 30, 40, 50]
nums.insert(0, 5)
# All 5 existing elements must shift right to make room.
# Total cost: O(n), not amortized.
print(nums)   # [5, 10, 20, 30, 40, 50]
````

<strong> **Concept 4: amortized analysis — the average price of a rare expensive step**</strong>

---
<h6>
<strong> The general idea </strong>

Some operations are cheap most of the time and expensive occasionally. A dynamic array append is cheap when there's spare capacity (O(1)) and expensive when the array is full (O(n), because every element must be copied to a bigger block). If you look at any single append, it might be O(n). But if you look at the <b>total</b> cost of n appends, the expensive operations are rare enough that the average per append is O(1). This is <b>amortized analysis</b>: spreading the cost of rare expensive operations across all the cheap ones.

The key insight is that <b>the expensive step buys capacity for many future cheap steps</b>. Doubling when full means that after paying O(n) to copy n elements, the next n appends are all O(1) because there's n slots of spare room. So the O(n) cost is "paid for" by n subsequent O(1) appends. The average is O(1).

This is not the same as "average case" analysis. Average case is about random inputs. Amortized analysis is about the <b>total cost over a sequence of operations</b>. A single operation can be O(n), but over many operations the average is O(1).
</h6>

<h6>

The $120 payment buys access for all 365 days. The average is $0.33, even though no single day actually costs $0.33.

This is exactly how amortized analysis works. The rare expensive payment is spread across the many cheap days.
</h6>

<h6>
<strong> Example 3: the doubling effect </strong>

Here's the mathematical heart of amortized analysis. If a dynamic array doubles capacity when full, the resize costs are:

- 1st resize (at size 1): copy 1 element
- 2nd resize (at size 2): copy 2 elements
- 3rd resize (at size 4): copy 4 elements
- 4th resize (at size 8): copy 8 elements
- ...

The total copies for n appends: 1 + 2 + 4 + 8 + ... + n = 2n - 1.

That's O(n) total, which is O(1) per append on average. The rare expensive resizes add up to less than 2× the number of appends.

Contrast with growing by 1:

- 1st resize: copy 1
- 2nd resize: copy 2
- 3rd resize: copy 3
- ...
- nth resize: copy n

Total copies: 1 + 2 + 3 + ... + n = n(n+1)/2 ≈ n²/2. That's O(n²) total, which is O(n) per append on average. Much worse.
</h6>

````python
def count_copies(n: int, grow) -> int:
    """Simulate n appends. grow(capacity) gives the new capacity.
    Returns the total number of element copies caused by resizing."""
    capacity, size, copies = 1, 0, 0
    for _ in range(n):
        if size == capacity:
            copies += size
            capacity = grow(capacity)
        size += 1
    return copies

print(count_copies(1000, lambda c: c * 2))   # 1023: doubling
print(count_copies(1000, lambda c: c + 1))   # 499500: growing by 1
````

<b>for _ in range(n):</b> — simulate n appends.

<b>if size == capacity:</b> — the array is full. Must resize.

<b>copies += size</b> — copying all current elements. This is the expensive step.

<h6>
<strong> Example 4: this is not "average case" </strong>

A common confusion: "amortized" ≠ "average case."

<b>Average case</b> — averaged over random inputs. If you sort a random array, the average case is O(n log n) even though the worst case is O(n²).

The distinction matters. If a rare expensive step only happens on some inputs, that's average case. If it happens periodically regardless of input, that's amortized. Dynamic array resizes happen periodically (every doubling), regardless of what data you're appending.

Amortized is about sequences of operations on any input. Average case is about random inputs. Dynamic array resizes are periodic regardless of input, so the analysis is amortized, not average-case.

<b> 4. The expensive step buys capacity for many cheap steps. </b>

A resize copies n elements but frees up n slots of spare capacity. The next n appends are O(1). So the expensive step is "paid for" by the cheap steps that follow.

<b> 5. This pattern appears everywhere in algorithm design. </b>

Hash tables resize when they fill up (amortized O(1) insertion). Heaps sift elements up and down (amortized O(log n)). Union-Find paths get compressed (amortized almost O(1)). Amortized analysis is the tool that makes these guarantees meaningful.
</h6>

<strong> **pytest — the test runner**</strong>

<h6>
<strong> The general idea </strong>

Writing tests is how you know your code works — and more importantly, how you know it still works after you change something. A test runner finds your tests, runs each one, and reports pass or fail per test. <b>pytest</b> is Python's standard test runner: it's simple, it discovers tests automatically, and it works with plain <b>assert</b> statements. No classes, no inheritance, no boilerplate. Just functions named <b>test_*</b> containing <b>assert</b> statements.
</h6>

<h6>
<strong> How discovery works </strong>

pytest looks for files named <b>test_*.py</b> and functions named <b>test_*</b>. Anything inside <b>if __name__ == "__main__":</b> is ignored. If your test doesn't start with <b>test_</b>, it silently never runs — no error, no warning, just no result. That's the most common mistake.
</h6>

````python
# test_math.py — pytest finds and runs test_add automatically
def test_add():
    assert 1 + 1 == 2

def test_subtract():
    assert 5 - 3 == 2
````

<h6>
<strong> Line-by-line </strong>

<b>def test_add():</b> — the name starts with <b>test_</b>, so pytest discovers and runs it.

<b>assert 1 + 1 == 2</b> — if this is True, the test passes silently. If False, pytest prints the values on both sides and reports a failure.

<b>def test_subtract():</b> — another test. Discovered separately. Fails independently — one test's failure doesn't stop the others.

<b>To run:</b> <b>python -m pytest -v</b> from the project root. The <b>-v</b> means "verbose" — it prints each test's name and result.
</h6>

<h6>
<strong> Testing exceptions with `pytest.raises` </strong>

When you expect a function to raise an exception, you don't need <b>try</b>/<b>except</b>/<b>else</b>. <b>pytest.raises</b> is a context manager that passes if the exception happens and fails if it doesn't.
</h6>

````python
import pytest

def test_pop_from_empty():
    with pytest.raises(IndexError):
        DynamicArray().pop()
````

<h6>
<strong> Line-by-line </strong>

<b>with pytest.raises(IndexError):</b> — the block should raise <b>IndexError</b>. If it does, the test passes. If it doesn't, the test fails with "DID NOT RAISE IndexError".

<b>DynamicArray().pop()</b> — the code being tested. Popping from an empty array should raise <b>IndexError</b>. The <b>with</b> block captures the exception and asserts it happened.

<b>Why this is better than try/except/else:</b> three lines instead of six, and pytest understands what you're testing. If the exception happens, it's reported as a pass. If it doesn't, it's reported as a failure with a clear message.
</h6>

<h6>
<strong> Common mistakes </strong>

<b>Running pytest from the wrong folder.</b> If imports break, you're not in the project root. Use <b>python -m pytest</b> from the root, not <b>pytest</b> from a subfolder.

<b>Forgetting the `test_` prefix.</b> A function named <b>check_add()</b> is ignored. Always prefix with <b>test_</b>.

<b>Putting asserts under `__main__`.</b> pytest never runs <b>if __name__ == "__main__":</b> blocks. The tests must be at module level.
</h6>

JupyterLab is a browser-based environment where code runs in cells and output appears directly below each cell. <b>Launching Jupyter outside your venv.</b> If <b>import pandas</b> fails in Jupyter but works in VS Code, Jupyter isn't using your venv. Activate the venv before running <b>jupyter lab</b>.

<strong> **Facts, mistakes, and interview prep**</strong>

<h6>
<strong> Counterintuitive facts </strong>

<b>CPython does not double its lists.</b> It grows them by about 12.5% plus a small constant. The amortized argument works for <b>any</b> constant growth factor above 1, and doubling is just the easiest to analyze. The 1,023 copies for 1,000 appends from the concept note is a doubling calculation; real CPython uses a slightly different formula.

<b>A list of one million distinct ints takes roughly 36 MB.</b> 8 MB of references plus about 28 bytes per int object (Python ints are objects, not raw values). A NumPy int64 array of the same numbers takes 8 MB — no per-item object overhead, just raw 8-byte values. This is why pandas columns (which are NumPy arrays) are so much faster for numeric work.
</h6>

<h6>
<strong> Common mistakes even experienced developers make </strong>

<b>[[0] * n] * m</b> — copies the same inner list's reference m times. Use <b>[[0] * n for _ in range(m)]</b> instead.

<b>Calling insert(0, x) or pop(0) in a loop.</b> Both are O(n) per call, so a loop of n calls is O(n²). Use <b>append</b> and <b>pop()</b> (from the end) instead.

<b>Describing append as O(1) without saying amortized.</b> A single append can be O(n) when a resize happens. The correct phrasing is "amortized O(1)."
</h6>

<h6>
<strong> Common interview questions </strong>

<b>Why is list indexing O(1) but `x in list` O(n)?</b>
Indexing uses address arithmetic: `start + i * slot_size`. Membership testing scans every element.

<b>What does amortized O(1) mean, and why does doubling give it while +1 growth doesn't?</b>
Amortized O(1) means the total cost of n operations is O(n). Doubling copies 1+2+4+...+n = 2n elements total, so O(1) per append. +1 growth copies 1+2+3+...+n = n²/2 elements total, so O(n) per append.

<b>Why is `list.insert(0, x)` slow?</b>
Every element must shift right by one position to make room. That's O(n) per insert. A loop of n inserts is O(n²).

<b>What's the bug in `[[0] * 3] * 3`?</b>
All three rows are the same list object. Mutating one row mutates all three. The fix is `[[0] * 3 for _ in range(3)]`.

<b>How is a Python list different from a NumPy array in memory?</b>
A Python list holds references to Python objects (each with its own overhead). A NumPy array holds raw values in a contiguous block, no references. That's why NumPy is faster and uses less memory for numeric data.
</h6>

<h6>
<strong> Production example </strong>

Your pandas columns are NumPy arrays: true contiguous blocks of raw values, no per-item reference hop. That's why profiling a million-row column is fast — the same address arithmetic from Concept 1, without the reference indirection. Java's ArrayList, C++'s vector, and Go's slices all use the same grow-and-copy design. The concepts you learned today are the concepts behind every major collection type in every language.
</h6>
