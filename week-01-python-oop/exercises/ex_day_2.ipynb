<strong> Challenge 1: Classes & Objects — "The Coffee Machine" </strong>

**Goal**
<h6>
Build a CoffeeMachine class that models a real coffee machine.

<u> Instructions :</u>


Create a class CoffeeMachine with:

- __init__(self, brand, water_level, beans_level) — stores brand and current levels (in ml and grams)

- brew(self) — returns a message; reduces water by 50ml and beans by 20g only if there's enough of both; otherwise returns an error message

- refill_water(self, amount) — increases water level

- refill_beans(self, amount) — increases beans level

- status(self) — returns a string showing brand, water, and beans </h6>



```python
# __init__(self, brand, water_level, beans_level) — stores brand and current levels (in ml and grams)
class CoffeeMachine:
    def __init__(self, brand, water_level, beans_level):
        self.brand = brand
        self.water_level = water_level
        self.beans_level = beans_level

    def __repr__(self):
        return f'{self.brand} has | {self.water_level}.ml of water and {self.beans_level}.g of beans'    

    # brew(self) — returns a message; reduces water by 50ml and beans by 20g only if there's enough of both; 
    # otherwise returns an error message
    def brew(self):
        if self.water_level >= 50 and self.beans_level >= 20:
            self.water_level -= 50
            self.beans_level -= 20
            return f"Brewing a coffee... ☕| water_level = {self.water_level} and beans = {self.beans_level}"    
        return f'Not enough water or beans!'
        
    #  refill_water(self, amount) — increases water level
    def refill_water(self, amount: int):
        self.water_level += amount
        return f"water level has increased by {amount}.ml, now there is {self.water_level}.ml"  

    # refill_beans(self, amount) — increases beans level
    def refill_beans(self, amount: int):
        self.beans_level += amount
        return f"bean level has increased by {amount}.g, now there is {self.beans_level}.g"

    # status(self) — returns a string showing brand, water, and beans
    def status(self):
        return f"{self.brand} | Water: {self.water_level}ml | Beans: {self.beans_level}g"
```

```python
machine = CoffeeMachine("BrewMaster", 200, 100)
print(repr(machine))
print(machine.brew())        # "Brewing a coffee... ☕"
print(machine.status())      # "BrewMaster | Water: 150ml | Beans: 80g"
print(machine.brew())        # "Brewing a coffee... ☕" 
print(machine.brew())        # "Not enough water or beans!"
print(machine.status())      
machine.refill_water(100)
machine.refill_beans(50)
print(machine.status())      # "BrewMaster | Water: 200ml | Beans: 110g"
```

<strong>Challenge 2: Instance vs. Class Attributes — "The Employee Roster" </strong>

**Goal**
<h6>
Demonstrate the difference between shared and unique data.

<u>Instructions</u>

Create a class Employee with:

1, Class attributes:

- company = "TechCorp"

- employee_count = 0

- standard_work_week = 40

2, Instance attributes (in __init__):

- name, role, salary

- hours_worked (starts at 0)

3, Methods:

- work(self, hours) — adds to hours_worked

- get_info(self) — returns name, role, salary, and whether they worked overtime (hours > standard_work_week)

4, Each new Employee should increment employee_count.</h6>

```python
class Employee:
    company = "TechCrop"
    employee_count = 0
    standard_work_week = 40

    def __init__(self, name: str, role: str, salary: float, hours_worked = 0):
        self.name = name
        self.role = role
        self.salary = salary
        self.hours_worked = hours_worked
        Employee.employee_count += 1

    def work(self, hours):
        self.hours_worked += hours 

    # whether they worked overtime (hours > standard_work_week)
    def is_overtime(self) -> bool:
        return self.hours_worked > Employee.standard_work_week    

    def get_info(self):
        return f"Employee_info = {self.name} | {self.role} | {self.salary} |overtime = {self.is_overtime()}"


e1 = Employee("Alice", "Engineer", 90000)
e2 = Employee("Bob", "Designer", 80000)
e3 = Employee("Charlie", "Manager", 100000)

print(f"total employee count = {Employee.employee_count}")   # 2 (shared)
print(f"company = {Employee.company}")          # "TechCorp" (shared)
print("*"*20)
e1.work(45)
e2.work(30)
e3.work(60)
print(e1.get_info())             # "Alice (Engineer) - $90000 | Overtime: True"
print(e2.get_info())             # "Bob (Designer) - $80000 | Overtime: False"
print(e3.get_info())             # "Charlie (Manager) - $100000 | Overtime: False"
print("*"*20)

# Change company for everyone
Employee.company = "NewCorp"
print(f"company = {Employee.company}")          # "NewCorp"
print(f"company = {e1.company}")                # "NewCorp"
print(f"company = {e2.company}")                # "NewCorp"
```

<strong> Challenge 3: Inheritance — "The Vehicle Family"</strong>

**Goal**

<h6> Build a small vehicle hierarchy that reuses code properly.

<u> Instructions </u>

1, Create a base class Vehicle with:

- __init__(self, brand, year, price)

- info(self) — returns "brand (year) - $price"

- age(self, current_year) — returns how old the vehicle is

2, Create three child classes:

- Car(Vehicle) — adds fuel_type; overrides info() to include fuel type

- Motorcycle(Vehicle) — adds engine_cc; overrides info() to include engine size

- Truck(Vehicle) — adds payload_tons; overrides info() to include payload

3, Each child must use super().__init__() and super().info(). </h6>

```python
class Vehicle:
    def __init__(self, brand, year, price):
        self.brand = brand
        self.year = year
        self.price = price

    def info(self):
        return f"{self.brand} ({self.year}) - {self.price}$"

    def age(self, current_year):
        age = current_year - self.year
        return f"age = {age} years"

class Car(Vehicle):  
    def __init__(self, brand, year, price, fuel_type):
        super().__init__(brand, year, price)
        self.fuel_type = fuel_type 
          
    def info(self):
        return f"{super().info()} | fuel: {self.fuel_type}"

class Motorcycle(Vehicle):
    def __init__(self, brand, year, price, engine_cc):
        super().__init__(brand, year, price)
        self.engine_cc = engine_cc
          
    def info(self):
        return f"{super().info()} | engine: {self.engine_cc}cc"

class Truck(Vehicle):
    def __init__(self, brand, year, price, load_tons):
        super().__init__(brand, year, price)
        self.load_tons = load_tons
          
    def info(self):
        return f"{super().info()} | load: {self.load_tons} tons"

car = Car("Toyota", 2020, 25000, "Petrol")
bike = Motorcycle("Harley", 2019, 15000, 1200)
truck = Truck("Volvo", 2018, 60000, 10)

print(car.info())     # "Toyota (2020) - $25000 | Fuel: Petrol"
print(bike.info())    # "Harley (2019) - $15000 | Engine: 1200cc"
print(truck.info())   # "Volvo (2018) - $60000 | Payload: 10 tons"

print(car.age(2026))  # 6
print(bike.age(2026)) # 7
print(truck.age(2026)) # 8
```

<strong> Challenge 4: Polymorphism — "The Shape Calculator" </strong>

**Goal**

<h6> Write code that treats different shapes uniformly.

<u> Instructions </u>

1, Create a base class Shape with:

- area(self) — raises NotImplementedError

- perimeter(self) — raises NotImplementedError

- describe(self) — returns "A shape"

2, Create three subclasses:

- Circle(radius) — area = πr², perimeter = 2πr

- Rectangle(width, height) — area = w·h, perimeter = 2(w+h)

- Triangle(a, b, c) — use Heron's formula for area, perimeter = a+b+c

3, Each overrides describe() to name itself. </h6>

```python
import math

class Shape:
    def __init__(self):
        self = self

    def total_area(self):
        raise NotImplementedError("Subclasses must implement this method")

    def total_perimeter(self):
        raise NotImplementedError("Subclasses must implement this method")        

    def describe(self):
        return f"This is a shape"

class Circle(Shape):            
    def __init__(self, radius):
        self.radius = radius

    def total_area(self):
        area = math.pi * self.radius ** 2
        return f"Circle area = {area:.2f}"

    def total_perimeter(self):
        perimeter = 2 * math.pi * self.radius
        return f"Circle perimeter = {perimeter:.2f}"

    def describe(self):
        return f"This is a Circle"    

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width        
        self.height = height

    def total_area(self):
        area = self.width * self.height
        return f"Rectangle area = {area:.2f}"

    def total_perimeter(self):
        perimeter = 2 * (self.width + self.height)
        return f"Rectangle perimeter = {perimeter:.2f}"

    def describe(self):
        return f"This is a Rectangle"    

class Triangle(Shape): 
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def total_area(self):
        # heron's formula
        s = (self.a + self.b + self.c) / 2
        area = math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))
        return f"Triangle area = {area:.2f}"

    def total_perimeter(self):
        perimeter = self.a + self.b + self.c
        return f"Triangle perimeter = {perimeter:.2f}"

    def describe(self):
        return f"This is a Triangle" 

shapes = [Circle(3), Rectangle(4, 5), Triangle(3, 4, 5)]

for shape in shapes:
    print('*'* 20)
    print(shape.describe())
    print(shape.total_area()) 
    print(shape.total_perimeter())  

def total_area(shapes):
    # do not use isinstance 
    return sum(shape.total_area() for shape in shapes)

for shape in shapes:
    print(shape.describe())
```

<strong> Challenge 6: Dunder Methods — "The Playlist Container" </strong>

**Goal**
<h6>
Make a custom class behave like a Python collection.

<u> Instructions </u>

1, Create a Playlist class with:

- __init__(self) — internal _songs list

- add(self, song) — appends a song (song is a string)

- __len__ — number of songs

- __contains__ — check if a song title exists

- __iter__ — allow for song in playlist

- __getitem__ — allow playlist [0] </6>

```python
class Playlist:
    def __init__(self):
        self._internal_songs = []

    def add(self, song):
        self._internal_songs.append(song)  

    def __len__(self):
        return len(self._internal_songs)

    def __contains__(self, song):
        return song in self._internal_songs

    def __iter__(self):
        return iter(self._internal_songs)  

    def __getitem__(self, song):
        return self._internal_songs[song]               

pl = Playlist()
pl.add("Song A")
pl.add("Song B")
pl.add("Song C")

print(len(pl))               # 3
print("Song A" in pl)        # True
print("Song Z" in pl)        # False

for song in pl:
    print(song)              # Song A, Song B, Song C

print('*'*20)
print(pl[1])                 # "Song B"
print(pl[-1])                # "Song C"
```

<strong> Final Capstone Challenge: "The Library Management System" </strong>

**Goal**
<h6>
Combine all six concepts into one cohesive system.

<u> Instructions</u>

Build a small library system with the following structure.

1. Base Class: LibraryItem

- Class attributes:

- library_name = "City Library"

- total_items = 0

- Instance attributes (via __init__):

- item_id, title, year, available (default True), checked_out_by (default None)

Methods:

- check_out(self, patron) — sets availability, returns a message

- return_item(self) — resets availability

- info(self) — returns a base description

- Increment total_items for every new item.

2. Subclasses

- Create three subclasses, each inheriting from LibraryItem:

- Book — adds author, pages; overrides info() to include them

- DVD — adds director, duration_minutes; overrides info() to include them

- Magazine — adds issue_number, publisher; overrides info() to include them

- Each must use super().__init__() and super().info().

3. Container Class: Library

- Internal list _items

- add(self, item) — appends an item

- __len__ — number of items

- __contains__ — check if a title exists in the library

- __iter__ — iterate over items

- __getitem__ — look up an item by item_id; raise KeyError if not found

4. Type-Aware Functions </h6>

```python
class LibraryItem:
    library_name = 'City Library'
    total_items = 0

    def __init__(self, item_id, title, year):
        self.item_id = item_id
        self.title = title
        self.year = year
        self.available = True
        self.checked_out_by = None
        LibraryItem.total_items += 1

    def check_out(self, patron):
        self.available = False
        self.checked_out_by = patron
        return f"{self.title} checkedout by to {patron}"
        
    def return_item(self):
        self.availabiity = True
        self.checked_out_by = None
        return f"{self.title} is available"

    def info(self):
        return f"id = {self.item_id} | title = {self.title} | year = {self.year} | available = {self.available} | checked_out_by_status = {self.checked_out_by}"   

class Book(LibraryItem):
    def __init__(self, item_id, title, year, author, pages):
        super().__init__(item_id, title, year)
        self.author = author
        self.pages = pages
    def info(self):
        return f"{super().info()} | author = {self.author} | {self.pages} pages"

class DVD(LibraryItem):
    def __init__(self, item_id, title, year, director, duration_minutes):
        super().__init__(item_id, title, year)  
        self.director = director
        self.duration_minutes = duration_minutes

    def info(self):
        return f"{super().info()} director = {self.director} | {self.duration_minutes} min"

class Magazine(LibraryItem):
    def __init__(self, item_id, title, year, issue_number, publisher):
        super().__init__(item_id, title, year)  
        self.issue_number = issue_number
        self.publisher = publisher

    def info(self):
        return f"{super().info()} is.no = {self.issue_number} | {self.publisher}"

class Library:
    def __init__(self):
        self._items = []

    def add(self, item):
        return self._items.append(item)

    def __len__(self):
        return len(self._items)

    def __contains__(self, title):
        return any(item.title == title for item in self._items)

    def __iter__(self):
        return iter(self._items)

    def __getitem__(self, item_id):
        for item in self._items:
            if (item.item_id == item_id):# for item in self._items)
                return item
        raise KeyError('item not found')        

def describe_item(item):
    if isinstance(item, Book):
        print(f'{item.info()}') 
    elif isinstance(item, DVD):
        print(f'{item.info()}') 
    elif isinstance(item, Magazine):
        print(f'{item.info()}')  
    else:
        print('not available')  

def list_all_items(Library):
    for item in Library:
        print(item.info())                                          
```

```python
lib = Library()

book = Book("B001", "Python 101", 2021, "Alice", 350)
dvd = DVD("D001", "Inception", 2010, "Nolan", 148)
mag = Magazine("M001", "Tech Today", 2024, 42, "TechPress")

lib.add(book)
lib.add(dvd)
lib.add(mag)

print(len(lib))                       # 3
print("Python 101" in lib)            # True
print("Unknown" in lib)               # False
print(lib["D001"].info())             # DVD info via __getitem__

print(book.check_out("Bob"))          # "Python 101 checked out to Bob"
print(book.info())                    # includes "Checked out by Bob"

describe_item(book)                   # "Book: ... | Available: False"
describe_item(dvd)                    # "DVD: ... | Available: True"

list_all_items(lib)                   # prints all three items' info

print(LibraryItem.total_items)        # 3
print(LibraryItem.library_name)       # "City Library"
```

```python
class Course:
    platform_name = 'LearnHub'
    total_courses = 0
    
    def __init__(self,course_id, title, price):
        self.course_id = course_id
        self.title = title
        self.price = price
        self.enrolled_students = 0
        Course.total_courses += 1

    def enroll(self):
        self.enrolled_students += 1

    def total_revenue(self) -> float:
        revenue = self.price * self.enrolled_students
        return revenue

    def summary(self):
        return f"{self.course_id} | {self.title} | ${self.price}"


class VideoCourse(Course):
    def __init__(self, course_id, title, price, instructor, duration_hours):
        super().__init__(course_id, title, price)
        self.instructor = instructor
        self.duration_hours = duration_hours

    def summary(self):
        return f"{super().summary()} | instructor: {self.instructor} | {self.duration_hours}h"

    def is_long(self):
        return  self.duration_hours >= 10   

class TextCourse(Course):
    def __init__(self, course_id, title, price, author, num_chapters):
        super().__init__(course_id, title, price)
        self.author = author
        self.num_chapters = num_chapters

    def summary(self):
        return f"{super().summary()} | Author: {self.author} | {self.num_chapters} chapters"

    def is_long(self) -> bool:
        return  self.num_characters >= 20

class Platform:
    def __init__(self):
        self._course = []

    def add(self, course):
        return self._course.append(course)

    def __len__(self):
        return len(self._course)

    def __contains__(self, course_id):
        return any(course.course_id == course_id for course in self._course)

    def __iter__(self):
        return iter(self._course)
    
    def __getitem__(self, course_id):
        for course in self._course:
            if (course.course_id == course_id):
                return course
        raise KeyError (f"No course with id {course_id!r}") 

def describe_course(course):
    if isinstance(course, VideoCourse):
        print(f"Video Course: {course.summary()}") 
    elif isinstance(course, TextCourse):
        print(f"Text Course: {course.summary()}")  
    else:
        print(f"Course: {course.summary()}")  
    
def total_revenue(Platform):
        return sum(course.total_revenue() for course in Platform)
```

```python
# Build platform
platform = Platform()

# Create courses
vc = VideoCourse("V001", "Python Mastery", 49.99, "Alice", 12.5)
tc = TextCourse("T001", "OOP Guide", 19.99, "Bob", 25)
short_vc = VideoCourse("V002", "Quick Intro", 9.99, "Carol", 3.0)

platform.add(vc)
platform.add(tc)
platform.add(short_vc)

# Enroll students
vc.enroll()
vc.enroll()
vc.enroll()          # 3 students
tc.enroll()          # 1 student
short_vc.enroll()    # 1 student

# --- Tests ---

print(len(platform))
# Expected: 3

print("V001" in platform)
# Expected: True

print("V999" in platform)
# Expected: False

print(platform["V001"].summary())
# Expected: T001 | OOP Guide | $19.99 | Author: Bob | 25 chapters

print(f"{vc.total_revenue():.2f}")
# Expected: 149.97

print(f"{total_revenue(platform):.2f}")
# Expected: 179.95

print(Course.total_courses)
# Expected: 3

print(Course.platform_name)
# Expected: LearnHub

Course.platform_name = "LearnHub Pro"
print(vc.platform_name)
# Expected: LearnHub Pro

# --- describe_course ---

describe_course(vc)
# Expected:
# Video Course: V001 | Python Mastery | $49.99 | Instructor: Alice | 12.5h
#   Long course: True

describe_course(tc)
# Expected:
# Text Course: T001 | OOP Guide | $19.99 | Author: Bob | 25 chapters
#   Long course: True

describe_course(short_vc)
# Expected:
# Video Course: V002 | Quick Intro | $9.99 | Instructor: Carol | 3.0h
#   Long course: False

# --- issubclass checks ---

print(issubclass(VideoCourse, Course))
# Expected: True

print(issubclass(Course, VideoCourse))
# Expected: False

print(issubclass(Platform, Course))
# Expected: False

# --- KeyError test ---

try:
    platform["MISSING"]
except KeyError as e:
    print(e)
# Expected: "No course with id 'MISSING'"
```
